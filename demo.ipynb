### Classes
Advanced data organization, a 'blueprint for creating objects'
- Class
    - blueprint for a position
- Object
    - one specific position

```python
class Position:
    def __init__(
            self,
            pid:str,
            grade:str,
            salary:float,
            union:str = None
    ):
        self.pid = pid
        self.grade = grade
        self.salary = salary
        self.union = union
    
pos = Position('P0123456','22-NONE-2',100000.0)

print(pos.pid)
```

#### Attributes, Properties, and Methods
Additional functionality for classes
- Attributes
    - Variables associated with a class or an object
    - Instance attributes
        - belong to a specific object (instance)
    - Class attributes
        - belong to the whole class and are shared by all instances

```python
class Position:
    MAX_FTE:float = 1.0

    def __init__(
            self,
            pid:str,
            grade:str,
            salary:float,
            union:str = None
    ):
        self.pid = pid
        self.grade = grade
        self.salary = salary
        self.union = union

pos1 = Position('P0123456','22-NONE-2',100000.0)
pos2 = Position('P0123454','23-NONE-2',105000.0)

print(pos1.pid,pos2.pid)
print(pos1.MAX_FTE,pos2.MAX_FTE)
```

- Properties
    - similar to attributes, access through additional behind-the-scenes logic
    - Used to:
        - control data access
        - add validation
        - compute values dynamically
    

```python
class Position:
    MAX_FTE:float = 1.0

    def __init__(
            self,
            pid:str,
            grade:str,
            salary:float,
            union:str = None,
            fte:float = 1.0
    ):
        self.pid = pid
        self.grade = grade
        self.salary = salary
        self.union = union
        self.fte = fte

    @property
    def union_status(self):
        if self.union is not None:
            return 'Union'
        else:
            return 'Non-Union'
        
    
pos1 = Position('P0123456','22-NONE-2',100000.0)
print(pos1.union_status)
```

- Methods
    - Functions defined within a class
    - Define behavior (what the object can do)

```python
class Position:
    MAX_FTE:float = 1.0

    def __init__(
            self,
            pid:str,
            grade:str,
            salary:float,
            union:str = None,
            fte:float = 1.0
    ):
        self.pid = pid
        self.grade = grade
        self.salary = salary
        self.union = union
        self.fte = fte
    
    def is_grade24(self) -> bool:
        return self.grade == '24'
    
    def update_salary(self, update_amount:float) -> None:
        self.salary += update_amount

pos1 = Position('P0123456','22-NONE-2',100000.0)

print(pos1.is_grade24())

pos1.update_salary(50000)
print(pos1.salary)

# print(pos1.update_salary(50000))

```

Uses `raise` keyword

- In our custom class

```python
class Position:
    MAX_FTE:float = 1.0

    def __init__(
            self,
            pid:str,
            grade:str,
            salary:float,
            union:str = None,
            fte:float = 1.0
    ):
        self.pid = pid
        self.grade = grade
        self.salary = salary
        self.union = union
        self.fte = fte

    @property
    def union_status(self):
        if self.union is not None:
            return 'Union'
        else:
            return 'Non-Union'
    
    @property # when property is called, run this
    def fte(self) -> float:
        return self._fte

    @fte.setter # when value is assigned, run this
    def fte(self,value:float) -> None:
        if not 0 < value <= self.MAX_FTE:
            raise ValueError(f'FTE value must be less than {self.MAX_FTE}')
        else:
            self._fte = value
            
pos = Position('P0123456','22-NONE-2',100000.0,fte=2.0)
```
